```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

df = pd.read_csv('driver_race_pit_analysis_1994_2022.csv')

df["pit_decile"] = pd.qcut(df["pit_time_pct"], 10, labels=False)

agg = (
    df
    .groupby("pit_decile", as_index=False)
    .agg(
        posities_gewonnen=("positions_gained", "mean"),
        pit_mid=("pit_time_pct", "mean")
    )
)

fig = px.scatter(
    agg,
    x="pit_mid",
    y="posities_gewonnen",
    trendline="ols",
    trendline_color_override="#666666",
    labels={
        "pit_mid":            "Gemiddelde Pitstop Tijd (relatief t.o.v. snelste = 1.0)",
        "posities_gewonnen":  "Gemiddeld Aantal Gewonnen Posities"
    },
    title="Gemiddeld Aantal Gewonnen Posities per Pitstop-Deciel",
    template="plotly_white",
    width=780,
    height=600
)

fig.data[1].name = "Trendlijn"
fig.data[1].showlegend = True

fig.add_trace(
    go.Scatter(
        x=agg["pit_mid"],
        y=agg["posities_gewonnen"],
        mode="lines+markers",
        line=dict(color="#B22222", width=3),
        marker=dict(size=8, color="#8B0000", line=dict(width=2, color="white")),
        name="Deciel Gemiddelden"
    )
)

fig.update_layout(
    plot_bgcolor="white",
    paper_bgcolor="#FFFFFF",
    
    xaxis=dict(
        tickformat=".2f", 
        gridcolor="#E0E0E0",
        title_font=dict(size=14, color="#2F2F2F"),
        tickfont=dict(size=12, color="#2F2F2F")
    ),
    yaxis=dict(
        gridcolor="#E0E0E0",
        title_font=dict(size=14, color="#2F2F2F"),
        tickfont=dict(size=12, color="#2F2F2F")
    ),

    title=dict(
        font=dict(size=18, color="#2F2F2F"),
        x=0.5,
        xanchor="center"
    ),

    legend=dict(
        bgcolor="rgba(255,255,255,0.8)",
        bordercolor="#DC143C",
        borderwidth=1,
        font=dict(size=12, color="#2F2F2F")
    ),
    
    margin=dict(l=80, r=40, t=80, b=60)
)

fig.update_traces(
    marker=dict(
        size=10,
        color="#DC143C",
        line=dict(width=2, color="white")
    ),
    selector=dict(mode="markers")
)

fig.show()
```
