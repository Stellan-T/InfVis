```python
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# 1. Load & prep
df = pd.read_csv("driver_race_pit_analysis_1994_2022.csv")
df["Year"] = (
    df["raceId"]
      .str.split(pat="_", n=1)
      .str[-1]
      .astype(int)
)
df = df.dropna(subset=["pit_sec"])

# 2. Compute the mean pit time and positions gained per year
yearly = (
    df
    .groupby("Year", as_index=False)
    .agg({
        "pit_sec": "mean",
        "positions_gained": "mean"
    })
    .rename(columns={
        "pit_sec": "avg_pit_sec",
        "positions_gained": "avg_positions_gained"
    })
)

# 3. Create subplot with secondary y-axis
fig = make_subplots(specs=[[{"secondary_y": True}]])

# 4. Add pit time line (primary axis)
fig.add_trace(
    go.Scatter(
        x=yearly["Year"],
        y=yearly["avg_pit_sec"],
        mode="lines+markers",
        name="Pit Stop Tijd (secondes)",
        line=dict(color="#DC143C", width=4),
        marker=dict(size=8, color="#DC143C", symbol="circle", 
                   line=dict(color="white", width=2))
    ),
    secondary_y=False,
)

# 5. Add positions gained bars (secondary axis)
fig.add_trace(
    go.Bar(
        x=yearly["Year"],
        y=yearly["avg_positions_gained"],
        name="Posities toegenomen",
        marker_color="rgba(100, 100, 100, 0.6)",
        marker_line=dict(color="#444444", width=1),
        width=0.6
    ),
    secondary_y=True,
)

# 6. Update layout with F1 styling
fig.update_xaxes(
    title_text="Seizoen/Jaar",
    dtick=2,
    gridcolor="rgba(128,128,128,0.2)",
    showgrid=True
)

# Primary y-axis (pit time)
fig.update_yaxes(
    title_text="<b>Pit Stop Tijd (secondes)</b>",
    title_font_color="#DC143C",
    title_font_size=14,
    tickfont_color="#DC143C",
    tickfont_size=12,
    secondary_y=False,
    gridcolor="rgba(128,128,128,0.3)",
    showgrid=True,
    range=[27, 37]
)

# Secondary y-axis (positions gained)
fig.update_yaxes(
    title_text="<b>Posities toegenomen</b>",
    title_font_color="#444444",
    title_font_size=14,
    tickfont_color="#444444",
    tickfont_size=12,
    secondary_y=True,
    showgrid=False,
    range=[0, 6]
)

fig.update_layout(
    title={
        'text': "<b>Gemiddelde pitstoptijd vs. gemiddelde positieverbetering</b>",
        'x': 0.5,
        'xanchor': 'center',
        'font': {'size': 20, 'color': '#2F2F2F'}
    },
    plot_bgcolor="white",
    paper_bgcolor="white",
    width=800,
    height=550,
    margin=dict(l=100, r=100, t=100, b=80),
    legend=dict(
        orientation="h",
        yanchor="top",
        y=-0.15,
        xanchor="center",
        x=0.5,
        bgcolor="rgba(255,255,255,0.9)",
        bordercolor="#CCCCCC",
        borderwidth=1,
        font=dict(size=14)
    ),
    font=dict(family="Arial", size=13, color="#2F2F2F")
)

fig.show()
```
